# Código-fonte RPA (Pyhton)

### Expondo o n8n local com ngrok

Para que o Colab possa se conectar ao seu n8n local, você precisa expor o seu servidor local para a internet. Uma ferramenta comum para isso é o `ngrok`.

````python
# Instalar pyngrok para usar ngrok (embora você vá rodar o ngrok localmente, isso seria para um caso de teste)
!pip install pyngrok -q
````

1.  **No seu terminal local**, inicie o ngrok para expor a porta do seu n8n (geralmente 5678):
    ```bash
    ngrok http 5678
    ```
2.  O ngrok irá fornecer uma URL pública (por exemplo, `https://abcdef12345.ngrok.io`).
3.  **Atualize a variável `N8N_WEBHOOK`** abaixo com a URL que o ngrok lhe forneceu. Lembre-se de adicionar o caminho `webhook-test/Clientes` ao final da URL do ngrok.

````python
# Atualize esta variável com a URL pública do ngrok
# Exemplo: N8N_WEBHOOK = "https://SUA_URL_NGROK.ngrok-free.app/webhook-test/Clientes"
N8N_WEBHOOK = "http://localhost:5678/webhook-test/Clientes" # Altere esta linha com a URL do ngrok
````

Após atualizar `N8N_WEBHOOK` com a URL do ngrok, você pode re-executar a célula anterior (`MmL6DwgHG38u`) para tentar enviar os dados ao seu webhook n8n.

````python
import requests
from bs4 import BeautifulSoup

URL_PAGINA = "https://digitalinnovationone.github.io/dio-lab-assistente-investimentos-rpa-n8n/"
N8N_WEBHOOK = "https://appeasing-halt-starving.ngrok-free.dev/webhook-test/Clientes"

# 1) Baixar a página
html = requests.get(URL_PAGINA, timeout=30).text

# 2) Ler a tabela de clientes
soup = BeautifulSoup(html, "html.parser")
linhas = soup.select("#clientes tbody tr")
clientes = []
for linha in linhas:
    colunas = linha.find_all("td")
    cliente = {
        "nome":  colunas[0].get_text(strip=True),
        "email": colunas[1].get_text(strip=True),
        "saldo": colunas[2].get_text(strip=True),
        "perfil": colunas[3].get_text(strip=True),
    }

    clientes.append(cliente)
print("Clientes encontrados:", len(clientes))
print("Exemplo 1º cliente:", clientes[0] if clientes else "nenhum")
# 3) Envia pro n8n
payload = {"clientes": clientes}
resp = requests.post(
    N8N_WEBHOOK,
    json=payload,
    headers={"Content-type": "application/json"},
    timeout=60
)
print("Status:", resp.status_code)
print("Headers:", dict(resp.headers))
print("Resposta (primeiros 1500 chars):")
print(resp.text[:1500])
````
